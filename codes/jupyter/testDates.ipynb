```python
import xarray as xr
import numpy as np
import glob
from datetime import datetime, timedelta
import pandas as pd
```

```python
startDate = datetime(2019, 6, 1)
endDate = datetime(2020, 3, 31)
curDate = startDate

folder = f'/srv/cdx/hseo/Data/GLORYS12v1/PCF/'
    

while curDate <= endDate:
    #print(curDate)
    fname = f'{curDate.year:04d}/GLORYS12v1_dailyAvg_{curDate:%Y-%m-%d}.nc'
    readFname = glob.glob(folder + fname)
    ds = xr.open_dataset(readFname[0])
    timeInFile = ds.time.to_numpy()
    if len(timeInFile) > 1:
        print(f"more than one time snapshots in {timeInFile}")
        print(f"'{curDate.year:04d}-{curDate.month:02d}-{curDate.day:02d},'")
        continue
    thisDate = pd.to_datetime(timeInFile[0])
    if thisDate.year != curDate.year or thisDate.month != curDate.month or thisDate.day != curDate.day:
        #print(thisDate, curDate)
        print(f"'{curDate.year:04d}-{curDate.month:02d}-{curDate.day:02d},'")
    curDate += timedelta(days=1)
```
