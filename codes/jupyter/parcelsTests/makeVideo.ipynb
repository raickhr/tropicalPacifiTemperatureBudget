```python
import cv2
import glob
import os

# Path to the folder containing frames
frame_folder = './images/'  # change if needed

# Get all PNG files and sort them by filename
frames = sorted(glob.glob(os.path.join(frame_folder, 'frame_*.png')))

# Read the first frame to get dimensions
frame = cv2.imread(frames[0])
height, width, layers = frame.shape

# Define output video
out = cv2.VideoWriter(
    'animation_fixed.mp4',  # output filename
    cv2.VideoWriter_fourcc(*'mp4v'),  # codec
    60,  # frames per second (change as needed)
    (width, height)
)

# Write frames
for filename in frames:
    img = cv2.imread(filename)
    out.write(img)

# Release everything
out.release()
print("✅ Video saved as animation.mp4")
```
