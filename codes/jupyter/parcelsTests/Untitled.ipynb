```python
import numpy as np
import matplotlib.pyplot as plt

# Range of coupling strength alpha
alpha = np.linspace(0.5, 1.0, 500)

# Compute frequency and check for real values
omega_sq = alpha**2 - (3*alpha - 2)**2
mask = omega_sq > 0
alpha = alpha[mask]
omega = np.sqrt(omega_sq[mask])

# Neutral-curve function (Eq. ★)
def delta_n(alpha, omega, n):
    phase = np.arccos((3*alpha - 2)/alpha)
    return (phase + 2*np.pi*n) / omega

# Plot several branches n = 0…3
plt.figure(figsize=(6,4))
for n in range(1):
    plt.plot(delta_n(alpha, omega, n), alpha,  label=f"n={n}")

plt.ylabel(r'$\alpha$ (coupling strength)')
plt.xlabel(r'$\delta$ (delay time)')
plt.title('Neutral curves of the delayed oscillator (Suarez & Schopf 1988)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
