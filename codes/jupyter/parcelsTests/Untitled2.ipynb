```python
gridDS = xr.open_dataset(folder + "glorysGrid.nc")
gridDS = gridDS.rename({'latitude':'gphif',
                        'longitude':'glamf'})
firstFile = ufiles[0]
ds = xr.open_dataset(firstFile)
gridDS['depthw'] = ds['depth']
gridDS.to_netcdf(folder + 'coordinates.nc')
```
